```python
import math
import itertools
import numpy as np
import pandas as pd
import sympy as sp
from math import comb
from collections import Counter


print("APM111-FA5\nRachelle Erika D. Borja")

# QUESTION 1: An experiment consists of drawing three cards in succession from a well-shuffled ordinary deck of cards.

data = {
    "Expression": [
        "P(E1 E2)",
        "P(E1 + E2)",
        "E1 + E2",
        "P(E3 | E1E2)",
        "E1 E2 E3",
        "P(E1E2 + E2E3)"
    ],
    "Meaning": [
        "Probability that a king appears on the first AND second draw",
        "Probability that a king appears on the first OR second draw",
        "Event that a king appears on the first OR second draw",
        "Probability that the third draw is a king given first two are kings",
        "Event that all three draws are kings",
        "Probability that (first & second are kings) OR (second & third are kings)"
    ]
}

df = pd.DataFrame(data)

print("\n" + "-" * 80 + "\n")
print("ANSWER TO Q1\n\nState in words the meaning of each of the following:")

display(df.style.hide(axis='index')
        .set_properties(**{'text-align': 'left'})
       )

print("\n" + "-" * 80 +"\n")


# QUESTION 2:

# All outcomes of three dice
outcomes = list(itertools.product(range(1, 7), repeat=3))
total_outcomes = len(outcomes)
sums = [sum(o) for o in outcomes]

# Frequency count
freq = Counter(sums)

# Probability distribution
values = sorted(freq.keys())
probabilities = [freq[v] / total_outcomes for v in values]

df = pd.DataFrame({
    "Sum (X)": values,
    "Probability P(X)": probabilities
})

display(df.style.hide(axis='index')
        .set_properties(**{'text-align': 'center'})
        .format({'Probability P(X)': '{:.4f}'}))

# Part (b)
print("\n(b) Computation of P(7 ≤ X ≤ 11)\n")

favorable = sum(freq[x] for x in range(7, 12))

print("Number of favorable outcomes =", favorable)
print("Total outcomes =", total_outcomes)

prob_range = favorable / total_outcomes

print("\nP(7 ≤ X ≤ 11) =",
      favorable, "/", total_outcomes,
      "=", prob_range)

print("\n" + "-" * 80 +"\n")

#QUESTION 3:
# Values and probabilities
x = np.array([-10, -20, 30])
p = np.array([1/5, 3/10, 1/2])

# Calculations
EX = np.sum(x * p)
EX2 = np.sum((x**2) * p)
variance = EX2 - EX**2
EX3 = np.sum((x**3) * p)

data = {
    "Quantity": [
        "E(X)",
        "E(X²)",
        "E[(X − μ)²]",
        "E(X³)"
    ],
    "Value": [
        EX,
        EX2,
        variance,
        EX3
    ]
}

df = pd.DataFrame(data)

print("ANSWER TO Q3\n\nGiven the probability distribution: Find: a, b, c, d")

display(df.style.hide(axis='index')
        .set_properties(**{'text-align': 'center'}, subset=['Value'])
        .set_properties(**{'text-align': 'left'}, subset=['Quantity'])
        .format({'Value': '{:.2f}'}))

print("\n" + "-" * 80 +"\n")

# QUESTION 4: In how many ways can 3 men and 3 women be seated at a round table if: (a), (b), (c)
# (a) No restriction imposed
a = math.factorial(6 - 1)

# (b) Two particular women must not sit together
total = math.factorial(5)
together = math.factorial(4) * 2
not_together = total - together
b = not_together

# (c) Each woman between two men
men_arrangements = math.factorial(3 - 1)
women_arrangements = math.factorial(3)
c = men_arrangements * women_arrangements

# Table Format
data = {
    "Case": [
        "(a) No restriction imposed",
        "(b) Two particular women must NOT sit together",
        "(c) Each woman is between two men"
    ],
    "Number of Ways": [a, b, c]
}

df = pd.DataFrame(data)

print("ANSWER TO Q4\n\nIn how many ways can 3 men and 3 women be seated at a round table if:")

display(df.style.hide(axis='index')
       .set_properties(**{'text-align': 'left'}, subset=['Case'])
       .set_properties(**{'text-align': 'center'}, subset=['Number of Ways'])
       )

print("\n" + "-" * 80 +"\n")


# QUESTION 5: In how many ways can 2 men, 4 women, 3 boys, and 3 girls be selected from: 6 men, 8 women, 4 boys, 5 girls
# (a) No restrictions
a = comb(6, 2) * comb(8, 4) * comb(4, 3) * comb(5, 3)

# (b) Particular man and woman must be selected
b = comb(5, 1) * comb(7, 3) * comb(4, 3) * comb(5, 3)

# Table Format
data = {
    "Case": [
        "(a) No restriction imposed",
        "(b) A particular man and woman must be selected"
    ],
    "Number of Ways": [a, b]
}

df = pd.DataFrame(data)

print("ANSWER TO Q5\n\nIn how many ways can 2 men, 4 women, 3 boys, and 3 girls be selected from:" + "\n" + "\t\t6 men, 8 women, 4 boys, 5 girls")

display(df.style.hide(axis='index')
       .set_properties(**{'text-align': 'left'}, subset=['Case'])
       .set_properties(**{'text-align': 'center'}, subset=['Number of Ways'])
       )

print("\n" + "-" * 80 +"\n")


# QUESTION 6: Prove that for any events E1 and E2, P (E1 + E2) ≤ P (E1) + P (E2)
# Defining symbolic probabilities
P1, P2, P12 = sp.symbols('P1 P2 P12', nonnegative=True)

# Addition Rule of Probability
P_Union = f"{P1} + {P2} - {P12}"

# Inequality for verification
inequality = sp.Le(P_Union, P1 + P2)

# Table Format
data = {
    "Expression": [
        "P(E₁ + E₂) using Addition Rule",
        "Right-hand side",
        "Inequality Result"
    ],
    "Value": [
        P_Union,
        P1 + P2,
        "P(E₁ + E₂) ≤ P(E₁) + P(E₂)"
    ]
}

df = pd.DataFrame(data)

print("ANSWER TO Q6\n\nProve that for any events E1 and E2, P (E1 + E2) ≤ P (E1) + P (E2)")
print("Using the Addition Rule of Probability, P(E₁ + E₂) = P(E₁) + P(E₂) - P(E₁E₂)" + "\n" + "Since P(E₁E₂) ≥ 0")

display(df.style.hide(axis='index')
       .set_properties(**{'text-align': 'left'}, subset=['Expression'])
       .set_properties(**{'text-align': 'center'}, subset=['Value'])
       )

print("\n" + "-" * 80)
```
