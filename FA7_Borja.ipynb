```python
import numpy as np
import pandas as pd
from scipy.stats import norm
from scipy.stats import t

print("APM 1111 - FA7\nBorja, Rachelle Erika D.")

# Problem 9.21
x = np.array([8.3, 10.6, 9.7, 8.8, 10.2, 9.4])
n = len(x)
x_bar = np.mean(x)

# Deviations and Squared Deviations
dev = x - x_bar
sq_dev = dev**2

# Sum of sq_dev
ssd = np.sum(sq_dev)

# Unbiased Sample Variance
s2 = ssd/(n-1)

# Sample Standard Deviation
s = np.sqrt(s2)

# Computation Table
table = pd.DataFrame({
    "Mass (kg)": x,
    "Deviation (x - x̄)": dev,
    "Squared Deviation (x - x̄)²": sq_dev
})

print("\n" + "-" * 80 + "\n")
print("Problem 9.21\n\nComputation Table\n")

display(
    table.style
    .format("{:.2f}")
    .hide(axis="index")
)

print("\nAnswer for 9.21(Summary of Results)")
print("Sample size (n):", n)
print("Sample mean (x̄):", round(x_bar,3))
print("Sum of squared deviations:", round(ssd,3))
print("Unbiased variance estimate (s²):", round(s2,3))
print("Sample standard deviation (s):", round(s,3))

print("\n" + "-" * 80 + "\n")


# **********************************************************
# Problem 9.24
mean = 11.09
s = 0.73
n = 60

# Standard Error
SE = s / np.sqrt(n)

# Z Critical Values
z95 = norm.ppf(0.975)
z99 = norm.ppf(0.995)

# Margins of Error
ME95 = z95 * SE
ME99 = z99 * SE

# Confidence Intervals
lower95 = mean - ME95
upper95 = mean + ME95

lower99 = mean - ME99
upper99 = mean + ME99

# Computation Table
table = pd.DataFrame({
    "Confidence Level": ["95%", "99%"],
    "Z Critical Value": [z95, z99],
    "Standard Error": [SE, SE],
    "Margin of Error": [ME95, ME99],
    "Lower Limit": [lower95, lower99],
    "Upper Limit": [upper95, upper99]
})

print("Problem 9.24\n\nConfidence Interval Computation Table\n")

display(
    table.style
    .format({
        "Z Critical Value": "{:.2f}",
        "Standard Error": "{:.2f}",
        "Margin of Error": "{:.2f}",
        "Lower Limit": "{:.2f}",
        "Upper Limit": "{:.2f}"
    })
    .hide(axis="index")
)

print("\nAnswer for 9.24(Summary of Results)")
print("Sample size (n):", n)
print("Sample mean:", round(mean,2))
print("Sample standard deviation:", round(s,2))
print("Standard Error:", round(SE,2))

print("\n95% Confidence Interval:")
print("(", round(lower95,2), ",", round(upper95,2), ")")

print("\n99% Confidence Interval:")
print("(", round(lower99,2), ",", round(upper99,2), ")")

print("\n" + "-" * 80 + "\n")

# **********************************************************
# Problem 9.33
nA = 50
nB = 100

meanA = 7.82
meanB = 6.75

sA = 0.24
sB = 0.30

# Difference of Means
diff = meanA - meanB

# Standard Error
SE = np.sqrt((sA**2)/nA + (sB**2)/nB)

# Z Critical Values
z95 = norm.ppf(0.975)
z99 = norm.ppf(0.995)

# Margins of Error
ME95 = z95 * SE
ME99 = z99 * SE

# Confidence Intervals
lower95 = diff - ME95
upper95 = diff + ME95

lower99 = diff - ME99
upper99 = diff + ME99

# Computation Table
table = pd.DataFrame({
    "Confidence Level": ["95%", "99%"],
    "Z Critical Value": [z95, z99],
    "Standard Error": [SE, SE],
    "Margin of Error": [ME95, ME99],
    "Lower Limit": [lower95, lower99],
    "Upper Limit": [upper95, upper99]
})

print("Problem 9.33\n\nConfidence Interval for Difference of Means\n")

display(
    table.style
    .format({
        "Z Critical Value": "{:.2f}",
        "Standard Error": "{:.2f}",
        "Margin of Error": "{:.2f}",
        "Lower Limit": "{:.2f}",
        "Upper Limit": "{:.2f}"
    })
    .hide(axis="index")
)

print("\nAnswer for 9.33(Summary of Results)")
print("Sample size A:", nA)
print("Sample size B:", nB)
print("Mean A:", meanA)
print("Mean B:", meanB)
print("Difference (A - B):", round(diff,2))
print("Standard Error:", round(SE,3))

print("\n95% Confidence Interval:")
print("(", round(lower95,2), ",", round(upper95,2), ")")

print("\n99% Confidence Interval:")
print("(", round(lower99,2), ",", round(upper99,2), ")")

print("\n" + "-" * 80 + "\n")

# **********************************************************
# Problem 9.34
# Male lifetimes
male = [
85,53,100,49,65,
60,51,61,83,65,
55,99,56,55,55,
90,72,62,69,59,
49,72,58,60,68,
90,74,85,80,77,
62,65,81,55,71,
78,49,78,80,75,
53,82,109,87,78,
72,104,70,31,50
]

# Female lifetimes
female = [
64,93,82,71,77,
64,60,75,87,60,
61,84,91,61,85,
105,90,59,86,62,
71,99,98,54,94,
98,61,108,79,50,
66,74,60,90,95,
81,86,65,86,81,
92,77,82,86,79,
91,93,63,93,53
]

male = np.array(male)
female = np.array(female)

# Sample Sizes, Means, & Standard Deviations
nM = len(male)
nF = len(female)

meanM = np.mean(male)
meanF = np.mean(female)

sM = np.std(male, ddof=1)
sF = np.std(female, ddof=1)

# Difference of Means
diff = meanM - meanF

# Standard Error
SE = np.sqrt((sM**2)/nM + (sF**2)/nF)

# Welch Degrees of Freedom
df = ((sM**2/nM + sF**2/nF)**2) / (
    ((sM**2/nM)**2)/(nM-1) + ((sF**2/nF)**2)/(nF-1)
)

# 85% Confidence Level
alpha = 0.15
tcrit = t.ppf(1 - alpha/2, df)

# Margin of Error
ME = tcrit * SE

# Confidence Interval
lower = diff - ME
upper = diff + ME

# Summary Table
summary = pd.DataFrame({
    "Group":["Males","Females"],
    "Sample Size":[nM,nF],
    "Mean":[meanM,meanF],
    "Std Dev":[sM,sF]
})

print("Problem 9.34\n\nSample Statistics\n")

display(
summary.style
.format({
    "Mean":"{:.2f}",
    "Std Dev":"{:.2f}"
})
.hide(axis="index")
)

# CI Table
ci = pd.DataFrame({
    "Confidence Level":["85%"],
    "Difference (Male-Female)":[diff],
    "Standard Error":[SE],
    "t Critical":[tcrit],
    "Margin of Error":[ME],
    "Lower Limit":[lower],
    "Upper Limit":[upper]
})

print("\nConfidence Interval Table\n")

display(
ci.style
.format({
    "Difference (Male-Female)":"{:.2f}",
    "Standard Error":"{:.2f}",
    "t Critical":"{:.2f}",
    "Margin of Error":"{:.2f}",
    "Lower Limit":"{:.2f}",
    "Upper Limit":"{:.2f}"
})
.hide(axis="index")
)
print("\nAnswer for 9.34(Result)")
print("85% Confidence Interval: (μ_MALE - μ_FEMALE)")
print("(", round(lower,2), ",", round(upper,2), ")")
```
